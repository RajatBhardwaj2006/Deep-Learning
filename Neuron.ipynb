```python
import torch
import torch.nn as nn
```

```python
# Input
torch.manual_seed(42) # random state
input = torch.tensor([1.0,2.0,3.0])
```

```python
neuron = nn.Linear(in_features=3, out_features=1)
```

```python
output = neuron(input)
```

```python
output
```

```python
# Verify output 
```

```python
print(f"Weight: {neuron.weight}\n")
print(f"Bias: {neuron.bias}")
```

```python
# using formula to verify

neuron.weight @ input + neuron.bias
```
